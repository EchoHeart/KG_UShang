```python
import pandas as pd
```

```python
order_data = pd.read_csv('原数据/订单信息.csv', encoding='GBK')
order_data
```

```python
order_data_hospital_1 = order_data[order_data['医院名称'].str.contains('宁波大学附属第一医院')]
order_data_hospital_1
```

```python
order_data_hospital_1_cleaned = order_data_hospital_1.copy()

# 统一“医院名称”字段
order_data_hospital_1_cleaned['医院名称'] = order_data_hospital_1_cleaned['医院名称'].str.replace('(', '（').str.replace(')', '）')
# 删除“护工ID”为空的记录
order_data_hospital_1_cleaned = order_data_hospital_1_cleaned.dropna(subset=['护工ID'])

order_data_hospital_1_cleaned['患者年龄'] = order_data_hospital_1_cleaned['患者年龄'].astype(int)
order_data_hospital_1_cleaned['护工ID'] = order_data_hospital_1_cleaned['护工ID'].astype(int)

order_data_hospital_1_cleaned
```

```python
# 异常值检测
order_data_hospital_1_cleaned[order_data_hospital_1_cleaned['患者年龄'] < 0]
```

```python
# 异常值检测
order_data_hospital_1_cleaned[order_data_hospital_1_cleaned['患者年龄'] > 104]
```

```python
order_data_hospital_1_cleaned.to_csv('订单信息（清洗后）_1.csv', index=False)
```
