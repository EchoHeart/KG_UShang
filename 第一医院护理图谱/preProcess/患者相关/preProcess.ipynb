```python
patientData = pd.read_csv('患者信息_3_宗教信仰已填充.csv')

# 填充学历
tmp = []
for i, row in patientData.iterrows():
    age = row['年龄']
    if age < 6:
        tmp.append('未上学')
    elif age <= 12:
        tmp.append('小学')
    elif age <= 15:
        tmp.append('初中')
    elif age <= 18:
        tmp.append('高中')
    else:
        arr = ['文盲', '小学', '初中', '高中', '大学及以上']
        if row['宗教信仰'] != '无宗教信仰':
            weights = [0.35, 0.4, 0.15, 0.1, 0]
        else:
            if age < 40:
                weights = [0, 0, 0.1, 0.3, 0.6]
            elif age > 60:
                weights = [0.4, 0.25, 0.2, 0.1, 0.05]
            else:
                weights = [0.1, 0.25, 0.3, 0.25, 0.1]
        flag = random.choices(range(5), weights=weights)[0]
        tmp.append(arr[flag])

patientData['学历'] = tmp
patientData.to_csv('患者信息_4_学历已填充.csv', index=False)
```

```python
patientData = pd.read_csv('患者信息_4_学历已填充.csv')

# 填充身高
tmp = []
for i, row in patientData.iterrows():
    if row['性别'] == '女':
        if row['年龄'] <= 30:
            tmp.append(random.choices(range(155, 170))[0])
        else:
            tmp.append(random.choices(range(150, 165))[0])
    else:
        if row['年龄'] >= 60:
            tmp.append(random.choices(range(160, 170))[0])
        elif row['年龄'] <= 30:
            tmp.append(random.choices(range(170, 185))[0])
        else:
            tmp.append(random.choices(range(165, 175))[0])

patientData['身高'] = tmp
patientData.to_csv('患者信息_5_身高已填充.csv', index=False)
```

```python
def BMI_evaluate(BMI):
    if BMI <= 18.4:
        return '偏瘦'
    elif BMI <= 23.9:
        return '正常'
    elif BMI <= 27.9:
        return '偏胖'
    else:
        return '肥胖'
```

```python
patientData = pd.read_csv('患者信息_5_身高已填充.csv')

# 填充体重、BMI指数、体型
weights = []
BMIs = []
bodies = []
tmp = [[185, 240], [180, 185], [240, 280], [280, 285]]  # 正常、偏瘦、偏胖、肥胖
for _, row in patientData.iterrows():
    if row['年龄'] <= 10:
        res = round(random.choices(range(185, 240))[0] / 10, 1)
    elif row['年龄'] >= 70:
        res = round(random.choices(range(180, 185))[0] / 10, 1)
    elif row['年龄'] >= 40:
        if row['性别'] == '女':
            flag = random.choices(range(4), weights=[0.2, 0.3, 0.4, 0.1])[0]
        else:
            flag = random.choices(range(4), weights=[0.2, 0.1, 0.4, 0.3])[0]
        res = round(random.choices(range(tmp[flag][0], tmp[flag][1]))[0] / 10, 1)
    else:
        if row['性别'] == '女':
            flag = random.choices(range(4), weights=[0.5, 0.3, 0.1, 0.1])[0]
        else:
            flag = random.choices(range(4), weights=[0.5, 0.2, 0.2, 0.1])[0]
        res = round(random.choices(range(tmp[flag][0], tmp[flag][1]))[0] / 10, 1)

    BMIs.append(res)
    bodies.append(BMI_evaluate(res))
    weights.append(round(res * (row['身高'] / 100) ** 2))

patientData['体重'] = weights
patientData['BMI指数'] = BMIs
patientData['体型'] = bodies
patientData.to_csv('患者信息_6_体重已填充.csv', index=False)
```

```python
patientData = pd.read_csv('患者信息_9_暴露史已填充.csv')

# 填充吸烟相关字段
isSmoke = []
smokeTime = []
smokeCnt = []
arr = ['不吸烟', '有吸烟史', '吸烟']
for _, row in patientData.iterrows():
    age = row['年龄']
    if age < 18:
        flag = 0
    elif age >= 40 and row['性别'] == '男':
        flag = random.choices(range(len(arr)), weights=[0.2, 0.4, 0.4])[0]
    else:
        flag = random.choices(range(len(arr)), weights=[0.6, 0.3, 0.1])[0]

    isSmoke.append(arr[flag])
    if flag == 0:
        smokeTime.append(0)
        smokeCnt.append(0)
    else:
        smokeCnt.append(round(random.uniform(1, 3), 1))
        if flag == 2:
            startSmokeAge = random.choices(range(12, age))[0]
            smokeTime.append(age - startSmokeAge)
        else:
            startSmokeAge = random.choices(range(12, age - 2))[0]
            endSmokeAge = random.choices(range(startSmokeAge + 1, age + 1))[0]
            smokeTime.append(endSmokeAge - startSmokeAge)

patientData['是否吸烟'] = isSmoke
patientData['烟龄'] = smokeTime
patientData['平均每日吸烟次数'] = smokeCnt
patientData.to_csv('患者信息_10_吸烟已填充.csv', index=False)
```

```python
patientData = pd.read_csv('患者信息_10_吸烟已填充.csv')

# 填充饮酒相关字段
isDrink = []
drinkCnt = []
drinkType = []
arr = ['不饮酒', '有饮酒史', '饮酒']
types = ['无', '白酒', '啤酒', '红酒', '果酒']
for _, row in patientData.iterrows():
    age, sex = row['年龄'], row['性别']
    if age < 18:
        flag = 0
    elif age >= 40 and sex == '男':
        flag = random.choices(range(len(arr)), weights=[0.2, 0.3, 0.5])[0]
    else:
        flag = random.choices(range(len(arr)), weights=[0.5, 0.3, 0.2])[0]

    isDrink.append(arr[flag])
    if flag == 0:
        drinkCnt.append(0)
        drinkType.append(types[0])
    else:
        drinkCnt.append(round(random.uniform(0.5, 2), 1))
        if age < 30:
            if sex == '男':
                flag = random.choices(range(len(types)), weights=[0, 0.1, 0.4, 0.2, 0.3])[0]
            else:
                flag = random.choices(range(len(types)), weights=[0, 0.1, 0.2, 0.3, 0.4])[0]
        else:
            if sex == '男':
                flag = random.choices(range(len(types)), weights=[0, 0.4, 0.3, 0.2, 0.1])[0]
            else:
                flag = random.choices(range(len(types)), weights=[0, 0.2, 0.3, 0.4, 0.1])[0]
        drinkType.append(types[flag])

patientData['是否饮酒'] = isDrink
patientData['平均每日饮酒次数'] = drinkCnt
patientData['常饮酒类'] = drinkType
patientData.to_csv('患者信息_11_饮酒已填充.csv', index=False)
```

```python
patientData = pd.read_csv('患者信息_11_饮酒已填充.csv')

# 填充锻炼频率
tmp = []
for _, row in patientData.iterrows():
    [sex, age, body] = row[['性别', '年龄', '体型']]
    if age >= 75 or age < 6:
        tmp.append(0)
    elif body == '肥胖':
        tmp.append(round(random.uniform(0, 0.1), 1))
    elif body == '偏胖':
        tmp.append(round(random.uniform(0, 0.2), 1))
    else:
        l, r = 0, 1
        if 65 <= age < 75:
            r = 0.2 if sex == '男' else 0
        if 55 <= age < 65:
            r = 0.3 if sex == '男' else 0.2
        if 45 <= age < 55:
            r = 0.4 if sex == '男' else 0.3
        if 35 <= age < 45:
            l = 0.2 if sex == '男' else 0.1
            r = 0.6 if sex == '男' else 0.4
        if 25 <= age < 35:
            l = 0.4 if sex == '男' else 0.2
            r = 0.8 if sex == '男' else 0.5
        if 15 <= age < 25:
            l = 0.5 if sex == '男' else 0.3
            r = 1 if sex == '男' else 0.6
        if 6 <= age < 15:
            r = 0.2
        tmp.append(round(random.uniform(l, r), 1))

patientData['锻炼频率'] = tmp
patientData.to_csv('患者信息_12_锻炼频率已填充.csv', index=False)
```

```python
patientData = pd.read_csv('患者信息_14_重症等级已填充.csv')
orderData = pd.read_csv('../../dataMyself/订单信息.csv', encoding='GBK')

patient2Order = {}
with open('../../dataMyself/患者_订单.txt', 'r') as file:
    for line in file.readlines():
        [patientID, orderList] = line.replace('\n', '').split('->')
        tmp = eval(orderList)
        patient2Order[patientID] = tmp
```

```python
# 填充所属医院，所在病区
hospitalList = []
areaList = []
for _, row in patientData.iterrows():
    orderIndex = patient2Order[str(row['患者ID'])][0]

    hospital = orderData.iloc[orderIndex]['医院名称']
    if '方桥' in hospital:
        hospitalList.append('FQ')
    elif '月湖' in hospital:
        hospitalList.append('YH')
    else:
        hospitalList.append('WT')

    area = orderData.iloc[orderIndex]['病区名称'].split(' ')[0]
    areaList.append(area)

patientData['所属医院'] = hospitalList
patientData['所在病区'] = areaList
patientData.to_csv('../../dataMyself/患者信息.csv', index=False)
```
