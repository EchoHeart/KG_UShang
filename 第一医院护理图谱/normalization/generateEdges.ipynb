```python
import pandas as pd
```

```python
nurseData = pd.read_csv('normalized_nurseData.csv')
nurseData
```

```python
tmp = {}
tokenId = 0

for col in ['擅长科室', '所属服务商']:
    for token in nurseData[col].unique().tolist():
        if token not in tmp.keys():
            tmp[token] = tokenId
            tokenId += 1

for i, row in nurseData.iterrows():
    for col in ['擅长科室', '所属服务商']:
        nurseData.loc[i, col] = tmp[row[col]]

nurseData
```

```python
min2Max = {}
columns = ['性别', '省', '民族', '宗教信仰', '学历', '普通话程度', '方言程度', '体型', '擅长科室', '所属服务商', '常驻医院', '主要负责病区', '其他负责病区']

for col in columns:
    values = nurseData[col].tolist()
    minValue = min(values)
    maxValue = max(values)

    min2Max[col] = (minValue, maxValue)

for i, row in nurseData.iterrows():
    for col in columns:
        (minValue, maxValue) = min2Max[col]
        nurseData.loc[i, col] = (row[col] - minValue) / (maxValue - minValue)

nurseData
```

```python
import numpy as np

def getSimilarity(list1, list2):
    return round(np.dot(list1, list2) / (np.linalg.norm(list1) * np.linalg.norm(list2)), 2)
```

```python
# nurseCount = nurseData.shape[0]
nurseCount = 200
```

```python
similarity_matrix = [[0] * nurseCount for _ in range(nurseCount)]
for i in range(nurseCount):
    for j in range(nurseCount):
        res = getSimilarity(nurseData.iloc[i][2:], nurseData.iloc[j][2:])
        if res >= 0.9:
            similarity_matrix[i][j] = res
np.array(similarity_matrix)
```

```python
with open('simi_matrix_nurse.txt', 'w') as file:
    for i in range(nurseCount):
        similarity_info = ', '.join(map(str, similarity_matrix[i]))
        file.write(f'{i}: {similarity_info}\n')
```

```python
k = 1
while k >= 0.8:
    edges = {}
    edgeCount = 0
    for i in range(nurseCount):
        tarList = []
        for j in range(nurseCount):
            if i != j:
                res = getSimilarity(nurseData.iloc[i][2:], nurseData.iloc[j][2:])
                if res >= k:
                    tarList.append(j)

        if len(tarList) > 0:
            edges[i] = tarList
            edgeCount += len(tarList)

    if edgeCount > 0:
        with open(f'edge/200_1000/edge_nurse_{round(k, 1)}.txt', 'w') as file:
            for i in range(nurseCount):
                if i in edges.keys():
                    edgeInfo = ', '.join(map(str, edges[i]))
                    file.write(f'{i}: {edgeInfo}\n')

            file.write(f'edgeCount: {edgeCount}\n')

    print(k, edgeCount)
    k -= 0.1
```
