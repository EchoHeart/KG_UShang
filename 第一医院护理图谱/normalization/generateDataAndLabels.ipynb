```python
vocabulary = {}
tokenId = 0

def addToVocabulary(v, t, i):
    if t not in v.keys():
        v[t] = i
        return True

    return False

for col in columnToken_common:
    for token in nurseData[col].unique().tolist():
        tokenId = tokenId + 1 if addToVocabulary(vocabulary, token, tokenId) else tokenId

    for token in patientData[col].unique().tolist():
        tokenId = tokenId + 1 if addToVocabulary(vocabulary, token, tokenId) else tokenId

print(vocabulary)
```

```python
for col in columnToken_nurse:
    for token in nurseData[col].unique().tolist():
        if token == '无其他负责病区':
            continue
        tokenId = tokenId + 1 if addToVocabulary(vocabulary, token, tokenId) else tokenId
vocabulary['无其他负责病区'] = tokenId
tokenId += 1
print(vocabulary)
```

```python
for col in columnToken_patient:
    for token in patientData[col].unique().tolist():
        tokenId = tokenId + 1 if addToVocabulary(vocabulary, token, tokenId) else tokenId

print(vocabulary)
```

```python
with open('token2Id.txt', 'w') as file:
    for k in vocabulary.keys():
        file.write(f'{k}: {vocabulary[k]}\n')
```

```python
nurseData_noToken = nurseData.copy()

for i, row in nurseData_noToken.iterrows():
    for col in columnToken_common + columnToken_nurse:
        nurseData_noToken.loc[i, col] = vocabulary[row[col]]

nurseData_noToken.to_csv('noToken_nurseData.csv', index=False)
```

```python
patientData_noToken = patientData.copy()

for i, row in patientData_noToken.iterrows():
    for col in columnToken_common + columnToken_patient:
        patientData_noToken.loc[i, col] = vocabulary[row[col]]

patientData_noToken.to_csv('noToken_patientData.csv', index=False)
```

```python
min2Max = {}
with open('minMax.txt', 'w') as file:
    for col in columnNumber_common:
        nurseList = nurseData_noToken[col].tolist()
        patientList = patientData_noToken[col].tolist()

        minValue = min(min(nurseList), min(patientList))
        maxValue = max(max(nurseList), max(patientList))

        file.write(f'{col}: {minValue}, {maxValue}\n')
        min2Max[col] = (minValue, maxValue)

    for col in columnNumber_nurse:
        nurseList = nurseData_noToken[col].tolist()
        minValue = min(nurseList)
        maxValue = max(nurseList)

        file.write(f'{col}: {minValue}, {maxValue}\n')
        min2Max[col] = (minValue, maxValue)

    for col in columnNumber_patient:
        patientList = patientData_noToken[col].tolist()
        minValue = min(patientList)
        maxValue = max(patientList)

        file.write(f'{col}: {minValue}, {maxValue}\n')
        min2Max[col] = (minValue, maxValue)
```

```python
nurseData_normalized = nurseData_noToken.copy()

for i, row in nurseData_normalized.iterrows():
    for col in columnNumber_common + columnNumber_nurse:
        (minValue, maxValue) = min2Max[col]
        nurseData_normalized.loc[i, col] = (row[col] - minValue) / (maxValue - minValue)

nurseData_normalized.to_csv('normalized_nurseData.csv', index=False)
```

```python
patientData_normalized = patientData_noToken.copy()

for i, row in patientData_normalized.iterrows():
    for col in columnNumber_common + columnNumber_patient:
        (minValue, maxValue) = min2Max[col]
        patientData_normalized.loc[i, col] = (row[col] - minValue) / (maxValue - minValue)

patientData_normalized.to_csv('normalized_patientData.csv', index=False)
```

```python
def evaluate_ability(df, positive):
    res = [0] * 4
    for c in columnGroup_nurse[1]:
        if c == '年龄':
            if df[c] <= 42:
                res[3] += 1
            elif df[c] <= 51:
                res[2] += 1
            elif df[c] <= 60:
                res[1] += 1
            else:
                res[0] += 1
        elif c == '学历':
            arr = ['文盲', '小学', '初中', '高中']
            res[arr.index(df[c])] += 1
        elif c == '普通话程度':
            res[df[c] + df['方言程度'] - 1] += 1
        elif c == '体型':
            arr = ['肥胖', '偏胖', '偏瘦', '正常']
            res[arr.index(df[c])] += 1
        elif c == '在岗时间':
            if df[c] <= 128:
                res[0] += 1
            elif df[c] <= 254:
                res[1] += 1
            elif df[c] <= 380:
                res[2] += 1
            else:
                res[3] += 1
        elif c == '擅长科室种类':
            if df[c] == 4:
                res[3] += 1
            else:
                res[df[c]] += 1
        elif c == '接单数':
            if df[c] <= 7:
                res[0] += 1
            elif df[c] <= 14:
                res[1] += 1
            elif df[c] <= 21:
                res[2] += 1
            else:
                res[3] +=1
        elif c == '活跃度':
            if df[c] < 0.1:
                res[0] += 1
            elif df[c] < 0.2:
                res[1] += 1
            elif df[c] < 0.3:
                res[2] += 1
            else:
                res[3] += 1
        elif c == '评分':
            if df[c] == 0:
                res[0] += 1
            elif df[c] == 1 or df[c] == 2:
                res[1] += 1
            elif df[c] == 3 or df[c] == 4:
                res[2] += 1
            else:
                res[3] += 1

    ans, maxVal = -1, -1
    for idx, val in enumerate(res):
        # 乐观主义
        if positive:
            if val >= maxVal:
                ans = idx
                maxVal = val
        else:
            if val > maxVal:
                ans = idx
                maxVal = val

    return ans
```

```python
from collections import Counter

ability_scores = []
for i in range(500):
    ability_scores.append(evaluate_ability(nurseData.iloc[i], positive=True))

Counter(ability_scores)
```

```python
def evaluate_health(df, positive):
    res = [0] * 4
    for c in columnGroup_patient[1]:
        if c == '年龄':
            if df[c] <= 25:
                res[3] += 1
            elif df[c] <= 50:
                res[2] += 1
            elif df[c] <= 75:
                res[1] += 2
            else:
                res[0] += 1
        elif c == '体型':
            arr = ['肥胖', '偏胖', '偏瘦', '正常']
            res[arr.index(df[c])] += 1
        elif c == '过敏史':
            if df[c] == '无过敏史':
                res[3] += 1
            elif df[c] == '室内灰尘过敏':
                res[0] += 1
            elif df[c] in ['动物毛发过敏', '化妆品类过敏']:
                res[1] += 1
            elif df[c] in ['花粉过敏', '鸡蛋过敏', '牛奶过敏']:
                res[2] += 1
        elif c == '是否残疾':
            if df[c] == '残疾':
                res[0] += 1
        elif c == '暴露史':
            if df[c] == '无暴露史':
                res[3] += 1
            elif df[c] == '粉尘':
                res[2] += 1
            elif df[c] == '化学品':
                res[1] += 1
            else:
                res[0] += 1
        elif c == '是否吸烟':
            if df[c] == '不吸烟':
                res[3] += 1
        elif c == '烟龄':
            if 0 < df[c] <= 23:
                res[2] += 1
            elif df[c] <= 46:
                res[1] += 1
            else:
                res[0] += 1
        elif c == '平均每日吸烟次数':
            if 0 < df[c] <= 1.6:
                res[2] += 1
            elif df[c] <= 2.3:
                res[1] += 1
            else:
                res[0] += 1
        elif c == '是否饮酒':
            if df[c] == '不饮酒':
                res[3] += 1
        elif c == '平均每日饮酒次数':
            if 0 < df[c] <= 0.9:
                res[2] += 1
            elif df[c] <= 1.4:
                res[1] += 1
            else:
                res[0] += 1
        elif c == '常饮酒类':
            if df[c] == '无常饮酒类':
                res[3] += 1
            elif df[c] == '白酒':
                res[0] += 1
            elif df[c] == '果酒':
                res[2] += 1
            else:
                res[1] += 1
        elif c == '锻炼频率':
            if df[c] == 0:
                res[0] += 1
            elif df[c] == 0.1:
                res[1] += 1
            elif df[c] >= 0.4:
                res[3] += 1
            else:
                res[2] += 1
        elif c == '刷牙频率':
            if df[c] <= 0.5:
                res[0] += 1
            elif df[c] <= 1:
                res[1] += 1
            elif df[c] <= 1.5:
                res[2] += 1
            else:
                res[3] += 1
        elif c == '重症等级':
            if df[c] == 1:
                res[3] += 1
            elif df[c] == 2:
                res[2] += 1
            elif df[c] == 3:
                res[1] += 1
            else:
                res[0] += 1

    ans, maxVal = -1, -1
    for idx, val in enumerate(res):
        # 乐观主义
        if positive:
            if val >= maxVal:
                ans = idx
                maxVal = val
        else:
            if val > maxVal:
                ans = idx
                maxVal = val

    return ans
```

```python
health_scores = []
for i in range(2500):
    health_scores.append(evaluate_health(patientData.iloc[i], positive=False))

Counter(health_scores)
```

```python
def evaluate(ability, health):
    if ability == 0:
        return health
    elif ability == 3:
        return 3 - health
    elif ability == 1:
        m = {
            0: 0,
            1: 1,
            2: 3,
            3: 2
        }
        return m[health]
    else:
        m = {
            0: 1,
            1: 3,
            2: 2,
            3: 0,
        }
        return m[health]
```

```python
column_common = ['性别', '年龄', '省', '民族', '宗教信仰', '学历', '身高', '体重', 'BMI指数', '体型']

with open('embeds_200_1000.txt', 'w') as file:
    for i in range(200):
        # 护工基本信息
        list1 = nurseData_normalized_enhanced.iloc[i][column_common]
        # 护工照护能力
        # ability_score = evaluate_ability(nurseData.iloc[i], positive=True)
        ability_score = ability_scores[i]
        # 护工空间信息
        spaceInfo_nurse = nurseData.iloc[i][columnGroup_nurse[2]]
        [hospital_nurse, primaryArea_nurse, otherArea_nurse] = spaceInfo_nurse

        for j in range(1000):
            """
            基本信息按相似度打标签
            """
            list2 = patientData_normalized_enhanced.iloc[j][column_common]
            similarity = round(np.dot(list1, list2) / (np.linalg.norm(list1) * np.linalg.norm(list2)), 2)

            if similarity < 0.25:
                a_score = 0
            elif similarity < 0.5:
                a_score = 1
            elif similarity < 0.75:
                a_score = 2
            else:
                a_score = 3

            # 患者健康状况
            # health_score = evaluate_health(patientData.iloc[i], positive=False)
            health_score = health_scores[j]
            b_score = evaluate(ability_score, health_score)

            """
            空间信息按规则树打标签
            """
            spaceInfo_patient = patientData.iloc[j][columnGroup_patient[2]]
            [hospital_patient, area_patient] = spaceInfo_patient

            if hospital_patient != hospital_nurse:
                c_score = 0
            else:
                if area_patient == primaryArea_nurse:
                    c_score = 3
                elif area_patient == otherArea_nurse:
                    c_score = 2
                else:
                    c_score = 1

            # # 匹配度
            # alpha = round((0.2 * a_score + 0.35 * b_score + 0.45 * c_score) / 3, 2)
            #
            # file.write(f'{i}-{j}: {alpha}\n')
            file.write(f'{i}-{j}: {a_score}, {b_score}, {c_score}\n')
            progress_percentage = (1000 * i + (j + 1)) / 2000
            print(f"进度: {progress_percentage:.2f}%", end="\r")
```
