```python
import random
```

```python
pairs, labels = [], []
with open('labels_500_2500.txt', 'r') as file:
    for line in file.readlines():
        line = line.replace('\n', '')
        [nurse2Patient, label] = line.split(': ')

        pairs.append(nurse2Patient)
        labels.append(label)
```

```python
count = 500 * 2500

train_index = int(count * 0.98)
val_index = int(count * 0.99)

indices = list(range(count))
random.shuffle(indices)

with open('train/trainLabels_500_2500_98.txt', 'w') as file:
    for i in indices[:train_index]:
        file.write(f'{pairs[i]}: {labels[i]}\n')

with open('val/valLabels_500_2500_1.txt', 'w') as file:
    for i in indices[train_index:val_index]:
        file.write(f'{pairs[i]}: {labels[i]}\n')

with open('test/testLabels_500_2500_1.txt', 'w') as file:
    for i in indices[val_index:]:
        file.write(f'{pairs[i]}: {labels[i]}\n')
```
