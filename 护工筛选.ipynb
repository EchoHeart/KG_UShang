```python
import pandas as pd
```

```python
caregiver_data = pd.read_csv('护工信息.csv', encoding='GBK')
caregiver_data
```

```python
# 去除“常驻医院”信息缺失的护工
caregiver_data_cleaned = caregiver_data.dropna(subset=['常驻医院'])
caregiver_data_cleaned
```

```python
# 筛选出“常驻医院”为宁波大学附属第一医院的护工
caregiver_data_hospital_1 = caregiver_data_cleaned[caregiver_data_cleaned['常驻医院'].str.contains('宁波大学附属第一医院')]
caregiver_data_hospital_1
```

```python
# 统一“常驻医院”中文小括号为英文小括号
caregiver_data_hospital_1_cleaned = caregiver_data_hospital_1.copy()
caregiver_data_hospital_1_cleaned['常驻医院'] = caregiver_data_hospital_1_cleaned['常驻医院'].str.replace('(', '（').str.replace(')', '）')

caregiver_data_hospital_1_cleaned
```

```python
caregiver_data_hospital_1_cleaned.to_csv('护工信息（筛选后）.csv', index=False)
```
