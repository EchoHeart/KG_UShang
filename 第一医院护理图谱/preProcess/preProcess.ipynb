```python
import pandas as pd
import numpy as np
import random
```

```python
nurseData = pd.read_csv('护工信息（推荐用）.csv')

# 填充评分
tmp = []
for _, row in nurseData.iterrows():
    if row['评分（满分5分）'] == '未知':
        tmp.append(random.choices(range(6), weights=[0.115, 0.2, 0.19, 0.195, 0.22, 0.08])[0])
    else:
        tmp.append(int(float(row['评分（满分5分）']) * 10) - 45)

nurseData['评分（满分5分）'] = tmp
nurseData.to_csv('护工信息_1_评分已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('护工信息_1_评分已填充.csv')

# 填充年龄
tmp = []

filtered = nurseData[nurseData['年龄'] != '未知']['年龄'].astype(int)
meanAge = filtered.mean()
stdAge = filtered.std()
for i, row in nurseData.iterrows():
    if row['年龄'] == '未知':
        tmp.append(int(np.random.normal(meanAge, stdAge)))
    else:
        tmp.append(int(row['年龄']))

nurseData['年龄'] = tmp
nurseData.to_csv('护工信息_2_年龄已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('护工信息_2_年龄已填充.csv')

# 填充省
tmp = []
arr = ['河南', '江苏', '浙江', '安徽', '贵州', '江西', '四川', '重庆']
for _, row in nurseData.iterrows():
    if row['省'] == '未知':
        flag = random.choices(range(8), weights=[0.48, 0.15, 0.16, 0.042, 0.042, 0.042, 0.042, 0.042])[0]
        tmp.append(arr[flag])
    else:
        tmp.append(row['省'][0:2])

nurseData['省'] = tmp
nurseData.to_csv('护工信息_3_省已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('护工信息_3_省已填充.csv')

# 填充民族
tmp = []
weights = [0.91] + [0.01] * 9
arr = ['汉族', '壮族', '回族', '满族', '维吾尔族', '苗族', '彝族', '土家族', '藏族', '蒙古族']
for _, row in nurseData.iterrows():
    if row['民族'] == '未知':
        flag = random.choices(range(10), weights=weights)[0]
        tmp.append(arr[flag])
    else:
        tmp.append(row['民族'])

nurseData['民族'] = tmp
nurseData.to_csv('护工信息_4_民族已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('护工信息_4_民族已填充.csv')

# 填充宗教信仰
tmp = []
weights = [0.95] + [0.01] * 5
arr = ['无宗教信仰', '道教', '佛教', '伊斯兰教', '天主教', '基督教']
for _, row in nurseData.iterrows():
    if row['宗教信仰'] == '未知':
        if row['民族'] == '藏族':
            flag = random.random()
            tmp.append('佛教' if flag <= 0.8 else '无宗教信仰')
        elif row['民族'] == '回族' or row['民族'] == '维吾尔族':
            flag = random.random()
            tmp.append('伊斯兰教' if flag <= 0.8 else '无宗教信仰')
        else:
            flag = random.choices(range(6), weights=weights)[0]
            tmp.append(arr[flag])
    else:
        tmp.append(row['宗教信仰'])

nurseData['宗教信仰'] = tmp
nurseData.to_csv('护工信息_5_宗教信仰已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('护工信息_5_宗教信仰已填充.csv', encoding='GBK')

# 填充学历
tmp = []
arr = ['文盲', '小学', '初中', '高中']
for _, row in nurseData.iterrows():
    if row['学历'] != '未知':
        tmp.append(row['学历'])
    else:
        if row['宗教信仰'] != '无宗教信仰':
            flag = random.random()
            tmp.append('文盲' if flag <= 0.8 else '小学')
        else:
            if row['评分（满分5分）'] >= 3:
                if row['年龄'] < 40:
                    flag = random.random()
                    tmp.append('高中' if flag <= 0.6 else '初中')
                else:
                    flag = random.choices(range(4), weights=[0.1, 0.31, 0.55, 0.04])[0]
                    tmp.append(arr[flag])
            else:
                if row['年龄'] < 40:
                    flag = random.random()
                    tmp.append('小学' if flag <= 0.6 else '初中')
                else:
                    flag = random.choices(range(4), weights=[0.7, 0.25, 0.05, 0])[0]
                    tmp.append(arr[flag])

nurseData['学历'] = tmp
nurseData.to_csv('护工信息_6_学历已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('护工信息_7_普通话程度已填充.csv')

# 填充宁波话程度
tmp = []
arr = ['不熟练', '能听懂', '能听懂能交流']
for _, row in nurseData.iterrows():
    if row['省'] == '浙江':
        tmp.append('能听懂能交流')
    else:
        if row['在岗时间'] >= 183:
            if row['评分（满分5分）'] >= 3:
                flag = random.choices(range(3), weights=[0, 0.2, 0.8])[0]
                tmp.append(arr[flag])
            else:
                flag = random.choices(range(3), weights=[0.3, 0.6, 0.1])[0]
                tmp.append(arr[flag])
        else:
            if row['评分（满分5分）'] >= 3:
                flag = random.choices(range(3), weights=[0, 0.4, 0.6])[0]
                tmp.append(arr[flag])
            else:
                flag = random.choices(range(3), weights=[0.8, 0.2, 0])[0]
                tmp.append(arr[flag])

nurseData['宁波话程度'] = tmp
nurseData.to_csv('护工信息_8_宁波话程度已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('护工信息_8_宁波话程度已填充.csv')
orderData = pd.read_csv('订单信息（画像用）.csv')

# 填充擅长科室
num = []
info = []
for _, row in nurseData.iterrows():
    df = orderData[orderData['护工ID'] == row['护工ID']]
    roomList = df['科室'].unique().tolist()
    if len(roomList) == 1:
        num.append(0)
        info.append('无')
    else:
        visited = []
        for room in roomList:
            if room != '未知':
                room = room if '科' in room else room[0:2] + '科'
                if room not in visited:
                    visited.append(room)

        num.append(len(visited))
        info.append('、'.join(visited))

nurseData['擅长科室'] = info
nurseData['擅长科室种类'] = num
nurseData.to_csv('护工信息_9_擅长科室已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('护工信息_9_擅长科室已填充.csv', encoding='GBK')

# 填充身高
tmp = []
for _, row in nurseData.iterrows():
    if row['身高'] != '未知':
        tmp.append(int(row['身高']))
    else:
        if row['性别'] == '女':
            tmp.append(random.choices(range(155, 165))[0])
        else:
            tmp.append(random.choices(range(165, 180))[0])

nurseData['身高'] = tmp
nurseData.to_csv('护工信息_10_身高已填充.csv', index=False)
```

```python
def BMI_evaluate(BMI):
    if BMI <= 18.4:
        return '偏瘦'
    elif BMI <= 23.9:
        return '正常'
    elif BMI <= 27.9:
        return '偏胖'
    else:
        return '肥胖'
```

```python
nurseData = pd.read_csv('护工信息_10_身高已填充.csv')

# 填充体重，生成BMI、身体状况
weights = []
BMIs = []
bodies = []
tmp = [[185, 240], [180, 185], [240, 280], [280, 285]]
for _, row in nurseData.iterrows():
    if row['体重'] != '未知':
        weight = int(row['体重'])
        weights.append(weight)
        res = round(weight / (row['身高'] / 100) ** 2, 1)
        BMIs.append(res)
        bodies.append(BMI_evaluate(res))
    else:
        if row['评分（满分5分）'] >= 4:
            res = round(random.choices(range(185, 240))[0] / 10, 1)
        elif row['评分（满分5分）'] <= 1:
            if row['性别'] == '女':
                flag = random.choices(range(4), weights=[0.2, 0.4, 0.24, 0.16])[0]
            else:
                flag = random.choices(range(4), weights=[0.2, 0.16, 0.24, 0.4])[0]
            res = round(random.choices(range(tmp[flag][0], tmp[flag][1]))[0] / 10, 1)
        else:
            if row['性别'] == '女':
                flag = random.choices(range(4), weights=[0.4, 0.36, 0.24, 0])[0]
            else:
                flag = random.choices(range(4), weights=[0.4, 0.24, 0.36, 0])[0]
            res = round(random.choices(range(tmp[flag][0], tmp[flag][1]))[0] / 10, 1)

        BMIs.append(res)
        bodies.append(BMI_evaluate(res))
        weight = round(res * (row['身高'] / 100) ** 2)
        weights.append(weight)

nurseData['体重'] = weights
nurseData['BMI指数'] = BMIs
nurseData['体型'] = bodies
nurseData.to_csv('护工信息_11_体重已填充.csv', index=False)
```

```python
nurseData = pd.read_csv('../dataMyself/护工信息_已脱敏.csv')

# 填充病区
mainArea = []
otherArea = []
for _, row in nurseData.iterrows():
    nurseID = row['护工ID']
    df = orderData[orderData['护工ID'] == nurseID]
    area2Cnt = {}
    for areaInfo in df['病区名称'].tolist():
        area = areaInfo.split(' ')[0]
        if area in area2Cnt.keys():
            area2Cnt[area] += 1
        else:
            area2Cnt[area] = 1

    if len(area2Cnt) == 1:
        mainArea.append(list(area2Cnt.keys())[0])
        otherArea.append('无')
    else:
        for i, item in enumerate(sorted(area2Cnt.items(), key=lambda x: x[1], reverse=True)):
            if i == 0:
                mainArea.append(item[0])
            if i == 1:
                otherArea.append(item[0])

nurseData['主要负责病区'] = mainArea
nurseData['其他负责病区'] = otherArea
nurseData.to_csv('../dataMyself/护工信息_已脱敏.csv', index=False)
```
