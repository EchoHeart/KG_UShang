```python
import pandas as pd
```

```python
orderData = pd.read_csv('dataMyself/订单信息.csv', encoding='GBK')
```

```python
areaID_list = orderData['病区ID'].unique().tolist()
areaID_list = [item[1] for item in sorted(enumerate(areaID_list), key=lambda x:x[1])]
```

```python
areaName_list, hospital_list = [], []

for areaID in areaID_list:
    [areaName, hospital] = orderData[orderData['病区ID'] == areaID].iloc[0][['病区名称', '医院名称']]
    areaName_list.append(areaName)
    hospital_list.append(hospital)

areaData = pd.DataFrame({
    '病区ID': areaID_list,
    '病区名称': areaName_list,
    '所属医院': hospital_list
})
```

```python
areaData.to_csv('病区数据.csv', index=False)
```

```python
import pandas as pd
import json
```

```python
data = pd.read_csv('dataMyself/订单信息.csv', encoding='GBK')
data
```

```python
def transformOrderID(orderID, createTime):
    [date, time] = createTime.split(' ')
    [year, month, day] = date.split('/')
    [hour, minute] = time.split(':')
    month = '0' + month if len(month) == 1 else month
    day = '0' + day if len(day) == 1 else day
    hour = '0' + hour if len(hour) == 1 else hour

    return "{:.0f}".format(orderID)[0:5] + year + month + day + hour + minute
```

```python
data_json = {}
for _, row in data.iterrows():
    ID = transformOrderID(row['订单号'], row['创建时间'])
    data_json[ID] = {}
    for col in data.columns[1:]:
        data_json[ID][col] = row[col]
```

```python
with open('order.json', 'w', encoding='utf-8') as file:
    file.write(json.dumps(data_json, indent=4, ensure_ascii=False))
```
