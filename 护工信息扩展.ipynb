```python
import pandas as pd
import numpy as np
```

```python
caregiver_data = pd.read_csv('护工信息（筛选后）.csv', encoding='GBK')
caregiver_data
```

```python
caregiver_extend = pd.read_csv('护工扩展信息.csv')
caregiver_extend
```

```python
extend_columns = list(caregiver_extend.columns)[2:]
```

```python
for col in extend_columns:
    tmp = []
    for i in range(caregiver_data.shape[0]):
        ID = caregiver_data.iloc[i]['护工ID']
        target = caregiver_extend[caregiver_extend['护工ID'] == ID]
        if target.shape[0] > 0:
            tmp.append(caregiver_extend[caregiver_extend['护工ID'] == ID][col].iloc[0])
        else:
            tmp.append(np.nan)
    caregiver_data[col] = tmp

caregiver_data
```

```python
caregiver_certification = pd.read_csv('护工资质信息.csv', encoding='GBK')
caregiver_certification
```

```python
cnt = 0
for i in range(caregiver_data.shape[0]):
    ID = caregiver_data.iloc[i]['护工ID']
    target = caregiver_certification[caregiver_certification['护工ID'] == ID]
    if target.shape[0] > 0:
        print(target)
        cnt += 1
print(cnt)
```
